```python
import numpy as np

latitude = np.array([  28.6139,28.6145,28.6152,28.6158,28.6164,28.6170])
longitude = np.array([77.2090,77.2098,77.2105,77.2112,77.2120,77.2122])

time = np.array([2, 2, 2, 2, 2])

dlat = np.diff(latitude)
dlon = np.diff(longitude)

distance = np.sqrt(dlat**2 + dlon**2) * 111

total_distance = np.sum(distance)

average_distance = np.mean(distance)

total_time_minutes = np.sum(time)
total_time_hours = total_time_minutes / 60

average_speed = total_distance / total_time_hours

remaining_distance = 2.5

total_trip_distance = total_distance + remaining_distance

remaining_time_hours = remaining_distance / average_speed
remaining_time_minutes = remaining_time_hours * 60

trip_progress = (total_distance / total_trip_distance) * 100

base_fare = 50
cost_per_km = 15

estimated_fare = base_fare + (total_trip_distance * cost_per_km)

if total_distance == 0:
    ride_status = "Driver Assigned"
elif remaining_distance > 0:
    ride_status = "Ride In Progress"
else:
    ride_status = "Ride Completed"


print("=" * 50)
print("              UBER GPS TRACKER")
print("=" * 50)

print("\nRIDE INFORMATION")
print("-" * 50)

print("Ride Status              :", ride_status)

print("Current Driver Location :",
      latitude[-1], ",", longitude[-1])

print("Pickup Location         :",
      latitude[0], ",", longitude[0])

print("Destination Distance    :",
      remaining_distance, "km")

print("\nGPS DISTANCE ANALYSIS")
print("-" * 50)

print("Distance between GPS updates:")
print(np.round(distance, 3), "km")

print("Average Distance per Update:",
      round(average_distance, 3), "km")

print("Total Distance Travelled :",
      round(total_distance, 3), "km")

print("Total Trip Distance      :",
      round(total_trip_distance, 3), "km")

print("\nTIME AND SPEED")
print("-" * 50)

print("Total Time Elapsed       :",
      total_time_minutes, "minutes")

print("Average Speed            :",
      round(average_speed, 2), "km/h")

print("\nETA INFORMATION")
print("-" * 50)

print("Remaining Distance       :",
      remaining_distance, "km")

print("Estimated Remaining Time :",
      round(remaining_time_minutes, 2), "minutes")

print("Estimated Arrival Time   :",
      round(remaining_time_minutes, 2), "minutes from now")

print("\nTRIP PROGRESS")
print("-" * 50)

print("Trip Progress            :",
      round(trip_progress, 2), "%")

print("\nFARE ESTIMATION")
print("-" * 50)

print("Base Fare                : Rs.", base_fare)
print("Cost Per KM              : Rs.", cost_per_km)
print("Estimated Ride Cost      : Rs.",
      round(estimated_fare, 2))

print("\n" + "=" * 55)
print("              LIVE GPS TRACKING")
print("=" * 55)

for i in range(len(distance)):

    print("\nGPS Update", i + 1)
    print("-" * 55)

    print("Latitude              :", latitude[i + 1])
    print("Longitude             :", longitude[i + 1])

    print("Distance Covered      :", round(distance[i], 3), "km")

    print("Time Taken            :", time[i], "minutes")

    print("Cumulative Distance   :",
          round(np.sum(distance[:i + 1]), 3), "km")

print("\n" + "=" * 55)
print("              CURRENT RIDE STATUS")
print("=" * 55)

print("Ride Status            :", ride_status)
print("Total Distance Travelled:",
      round(total_distance, 3), "km")

print("Average Speed          :",
      round(average_speed, 2), "km/h")

print("Distance Remaining     :",
      remaining_distance, "km")

print("Estimated Arrival      :",
      round(remaining_time_minutes, 2), "minutes")
print("Estimated Ride Cost    :",
      "Rs.", round(estimated_fare, 2))

print("\nTracking Status: ACTIVE")
```
